```python
#import the required modules
import cv2
import matplotlib.pyplot as plt
from deepface import DeepFace
```

```python
# read image
img = cv2.imread('happy.jpeg')
#img = cv2.imread('angry_man.jpeg')
#img = cv2.imread('sad.jpeg')
#img = cv2.imread('fear.jpeg')
#img = cv2.imread('surprise.jpeg')
#img = cv2.imread('neutral.jpeg')
#img = cv2.imread('disgust.jpeg')

# call imshow() using plt object
plt.imshow(img[:, :, : : -1])

# display that image
plt.show()
```

```python
# storing the result
result = DeepFace.analyze(img,
                          actions = ['emotion'])

# print result
print(result)
```

```python
# DeepFace may return list or dict depending on version
if isinstance(result, list):
    result = result[0]

# Get dominant emotion
dominant_emotion = result["dominant_emotion"]
emotion_scores = result["emotion"]

print("Dominant Emotion:", dominant_emotion)
print("\nEmotion scores:")
for emo, score in emotion_scores.items():
    print(f"{emo}: {score:.2f}%")
```
